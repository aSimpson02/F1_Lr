```python
#loading session
session = fastf1.get_session(2021, 7, 'Q')
session.name
session.date

#getting no.of laps
laps = session.laps

#no. of drivers 
drivers = session.drivers
print(drivers)
print('raw laps:', len(laps))


#cleaning data from laps 
clean = (laps.pick_wo_box()        # drop in/out laps
             .pick_track_status('1')  # green flag only
             .pick_quicklaps(1.07))   # drop laps >107% of fastest
clean = clean[clean['IsAccurate']].copy()
clean['LapSec'] = clean['LapTime'].dt.total_seconds()
print('clean laps:', len(clean))
clean[['Driver','LapNumber','Stint','Compound','TyreLife','LapSec']].head()
```
